# Fig. 5 · Amenity distance differentials

Each point uses the interquartile means of the paired home-to-POI distances for one of 63 amenity classes. The distance analysis excludes Park and Tourism.
Figure 5 shows the proportional difference, and the supplementary figure shows the difference in metres.
Marker area follows four bins of national POI counts.

Run all cells with the spatial kernel using the supplied `data/aggregated/amenity_distances.parquet`.

```python
from pathlib import Path
import sys

root = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "src" / "amenity_distance_plot.py").is_file())
sys.path.insert(0, str(root / "src"))

%matplotlib inline
import pandas as pd

from amenity_distance_plot import plot_distances
from figure_style import DATA_DIR, configure_style, export_figure

data = pd.read_parquet(DATA_DIR / "aggregated" / "amenity_distances.parquet")
configure_style("final")
```

## Fig. 5 · Proportional difference

```python
fig = plot_distances(data, "delta_d_rel", r"$\Delta d_{rel}$")
export_figure(fig, "fig5/fig5-amenity-distance-differentials", pad_inches=.03)
```

## Supplementary figure · Absolute difference

```python
fig = plot_distances(data, "delta_d", r"$\Delta d$ (m)")
export_figure(fig, "fig5/si-amenity-absolute-distance", pad_inches=.03)
```
